## Metoda Eulera nie-stabilna dla wahadla



Tworzymy układ dwóch równań pierwszego rzędu:
$$
\begin{cases} 
\frac{dx}{dt} = v \\ 
\frac{dv}{dt} = -\frac{g}{l}\sin(x) 
\end{cases}
$$

```julia
# Wahadło matematyczne – metoda Eulera
using Plots

# algorytm jest stabilny numeryczny jesli delta_t <= 2/(df/du), dlatego jesli wybierzemy
# dostatecznie male delta_t to algorytm jest stabilny, jesli natomiast ono bedzie za duze
# to algorytm bedzie niestabilny

function euler_pendulum(x0, v0, g, l, dt, t_end)
    N = Int(t_end / dt)
    t = zeros(Float64, N)
    x = zeros(Float64, N)
    v = zeros(Float64, N)

    x[1] = x0
    v[1] = v0

    for n in 1:N-1
        x[n+1] = x[n] + dt * v[n]
        v[n+1] = v[n] - dt * (g / l) * sin(x[n])
        t[n+1] = t[n] + dt
    end

    return t, x, v
end
```

#### Symulacja nie-stabilna - małe delta_t

### NOTE: metoda eulera daje bledny wynik - amplituda caly czas rosnie (co nie ma miejsca w wahadle) - jest to spodziewany efekt przy uzyciu tej metody
Zeby rozwiazac ten problem, trzeba na kazdym kroku dobierac odpowiednie delta_t w celu unikniecia niestabilnosci

```julia
# Parametry
g = 9.81         # Przyspieszenie ziemskie
l = 1.0          # Długość wahadła
x0 = π / 2       # Początkowe wychylenie
v0 = 0.0         # Początkowa prędkość
dt = 0.01        # Krok czasowy
t_end = 10.0     # Czas końcowy

# Symulacja
t, x, v = euler_pendulum(x0, v0, g, l, dt, t_end)

# Wykres
plot(t, x, xlabel="Czas [s]", ylabel="Wychylenie x(t) [rad]",
     title="Wahadło matematyczne — metoda Eulera", legend=false)
```

## Ulepszona metoda Eulera - metoda jawna




```julia
function euler_better_pendulum(x0, v0, g, l, dt, t_end)
    N = Int(t_end / dt)
    t = zeros(N)
    x = zeros(N)
    v = zeros(N)

    x[1] = x0
    v[1] = v0

    for n in 1:N-1
        # Krok pośredni w t + dt/2
        x_half = x[n] + (dt / 2) * v[n]
        v_half = v[n] - (dt / 2) * (g / l) * sin(x[n])

        # Krok pełny z użyciem wartości pośrednich
        x[n+1] = x[n] + dt * v_half
        v[n+1] = v[n] - dt * (g / l) * sin(x_half)
        t[n+1] = t[n] + dt
    end

    return t, x, v
end
```

### Stabilna - małe delta_t

```julia
using Plots
# Parametry
g = 9.81
l = 1.0
x0 = π / 2
v0 = 0.0
dt = 0.1
t_end = 10

# Symulacja
t, x, v = euler_better_pendulum(x0, v0, g, l, dt, t_end)

# Wykres
plot(t, x, xlabel="Czas [s]", ylabel="x(t) [rad]",
     title="Wahadło — ulepszona metoda Eulera (stabilna)", legend=false)
```

## Metoda Rungego-Kutty (4 stopnia)
Przyblizamy za pomoca szeregu Taylor'a. Decydujemy jak bardzo chcemy przyblizyc funkcję (jak daleko rozwinac szereg) i jak najmnijeszym kosztem obliczeniowym (oblicznie kolejnych pochodznych)



Wzór Rungego-Kutty dla r=4



```julia
using Plots

# eps ~ O(h^4)
# F(t,u,h) = c1*k1 + +c2*k2 + c3*k3

function runge_kutta_4_pendulum(x0, v0, g, l, h, t_end)
    N = Int(t_end / h)
    t = zeros(N)
    x = zeros(N)
    v = zeros(N)

    x[1] = x0
    v[1] = v0
    
    function f(u)
        return [u[2], - (g / l) * sin(u[1])]
    end


    for i in 1:N-1
        # u = [x, v], du = [v, -g/l * sin(x)]
        xi, vi = x[i], v[i]


        u = [xi, vi]

        k1 = f(u)
        k2 = f(u .+ 0.5h .* k1)
        k3 = f(u .+ 0.5h .* k2)
        k4 = f(u .+ h .* k3)

        u_next = u .+ (h/6) .* (k1 .+ 2k2 .+ 2k3 .+ k4)

        x[i+1] = u_next[1]
        v[i+1] = u_next[2]
        t[i+1] = t[i] + h
    end

    return t, x, v
end

# Parametry symulacji
g = 9.81
l = 1.0
x0 = π / 2
v0 = 0.0
h = 0.05
t_end = 20.0

# Symulacja
t, x, v = runge_kutta_4_pendulum(x0, v0, g, l, h, t_end)

# Wykres
plot(t, x, xlabel="Czas [s]", ylabel="x(t) [rad]",
     title="Wahadło — Runge-Kutta 4 rzędu",
     legend=false)
```

## Porownanie stabilnosci metod na podstawie wychylenia

```julia
# Parametry symulacji
g = 9.81
l = 1.0
x0 = π / 4
v0 = 0.0
t_end = 50.0
dts = [0.01, 0.05, 0.1, 0.2, 0.4, 0.5]
plots_list = []

for dt in dts
    t_e, x_e, _ = euler_pendulum(x0, v0, g, l, dt, t_end)
    t_ue, x_ue, _ = euler_better_pendulum(x0, v0, g, l, dt, t_end)
    t_rk, x_rk, _ = runge_kutta_4_pendulum(x0, v0, g, l, dt, t_end)
    
    p = plot(t_e, x_e, label="Euler", color=:red)
    plot!(p, t_ue, x_ue, label="Ulepszony Euler", color=:pink)
    plot!(p, t_rk, x_rk, label="RK4", color=:blue, 
          xlabel="Czas [s]", ylabel="Wychylenie [rad]", 
          title="Porównanie stabilności (dt = $dt)")
    push!(plots_list, p)
end

plot(plots_list..., layout=(2,3), size=(1600, 1000))

```

### Równania różniczkowe podwójnego wahadła (Double Pendulum)

Układ opisujący ruch podwójnego wahadła składa się z czterech równań różniczkowych pierwszego rzędu, gdzie:

* $\theta_1, \theta_2$ — kąty odchylenia ramion wahadła,
* $\omega_1, \omega_2$ — prędkości kątowe ($\omega_1 = \theta_1'$, $\omega_2 = \theta_2'$),
* $m_1, m_2$ — masy ramion,
* $L_1, L_2$ — długości ramion.


$$
\begin{aligned}
\theta_1' &= \omega_1 \\
\theta_2' &= \omega_2 \\
\omega_1' &= \frac{-g (2 m_1 + m_2) \sin \theta_1 - m_2 g \sin(\theta_1 - 2 \theta_2) - 2 \sin(\theta_1 - \theta_2)\, m_2 (\omega_2^2 L_2 + \omega_1^2 L_1 \cos(\theta_1 - \theta_2))}{L_1 (2 m_1 + m_2 - m_2 \cos(2 \theta_1 - 2 \theta_2))} \\
\omega_2' &= \frac{2 \sin(\theta_1 - \theta_2)\left(\omega_1^2 L_1 (m_1 + m_2) + g(m_1 + m_2) \cos \theta_1 + \omega_2^2 L_2 m_2 \cos(\theta_1 - \theta_2)\right)}{L_2 (2 m_1 + m_2 - m_2 \cos(2 \theta_1 - 2 \theta_2))}
\end{aligned}
$$


```julia
# Modyfikacja RK4 z poprzedniego zadania, zeby dzialalo na podwojnym wahadle

function double_pendulum_derivatives(t, theta1, omega1, theta2, omega2, g, l1, l2, m1, m2)
    # theta - kąt miedzy wahadlem a stanem rownowagi
    # omega - predkosc kotawa
    delta = theta1 - theta2
    denom = 2 * m1 + m2 - m2 * cos(2 * delta)

    dtheta1 = omega1
    dtheta2 = omega2

    domega1 = (
        -g * (2 * m1 + m2) * sin(theta1) - m2 * g * sin(theta1 - 2 * theta2) -
        2 * sin(delta) * m2 * (omega2^2 * l2 + omega1^2 * l1 * cos(delta))
    ) / (l1 * denom)

    domega2 = (
        2 * sin(delta) * (
            omega1^2 * l1 * (m1 + m2) + g * (m1 + m2) * cos(theta1) + omega2^2 * l2 * m2 * cos(delta)
        )
    ) / (l2 * denom)

    return dtheta1, domega1, dtheta2, domega2
end

function rk4_double_pendulum(tspan, dt, theta1_0, omega1_0, theta2_0, omega2_0, g, l1, l2, m1, m2)
    t_steps = collect(tspan[1]:dt:tspan[2])
    n = length(t_steps)

    theta1 = zeros(n)
    omega1 = zeros(n)
    theta2 = zeros(n)
    omega2 = zeros(n)

    theta1[1] = theta1_0
    omega1[1] = omega1_0
    theta2[1] = theta2_0
    omega2[1] = omega2_0

    for i in 1:n-1
        t = t_steps[i]
        y = [theta1[i], omega1[i], theta2[i], omega2[i]]

        function f(y)
            dtheta1, domega1, dtheta2, domega2 = double_pendulum_derivatives(t, y[1], y[2], y[3], y[4], g, l1, l2, m1, m2)
            return [dtheta1, domega1, dtheta2, domega2]
        end

        k1 = dt * f(y)
        k2 = dt * f(y .+ 0.5 .* k1)
        k3 = dt * f(y .+ 0.5 .* k2)
        k4 = dt * f(y .+ k3)

        y_next = y .+ (1/6) .* (k1 .+ 2k2 .+ 2k3 .+ k4)

        theta1[i+1], omega1[i+1], theta2[i+1], omega2[i+1] = y_next
    end

    return t_steps, theta1, theta2
end

function animate_double_pendulum(theta1, theta2, l1, l2; save_as="double_pendulum.gif", fps=60)
    n = length(theta1)

    x2_all = [l1 * sin(theta1[i]) + l2 * sin(theta2[i]) for i in 1:n]
    y2_all = [-l1 * cos(theta1[i]) - l2 * cos(theta2[i]) for i in 1:n]

    anim = @animate for i in 1:5:n
        x1 = l1 * sin(theta1[i])
        y1 = -l1 * cos(theta1[i])
        x2 = x1 + l2 * sin(theta2[i])
        y2 = y1 - l2 * cos(theta2[i])

        # wahadlo
        p1 = plot([0, x1], [0, y1], lw=3, color=:blue,
                  xlim=(-(l1+l2), l1+l2), ylim=(-(l1+l2), 1),
                  legend=false, aspect_ratio=1, title="Double Pendulum")
        plot!(p1, [x1, x2], [y1, y2], lw=3, color=:green)
        scatter!(p1, [x1, x2], [y1, y2], color=[:blue :red])

        # trajektoria konca wahadla
        p2 = plot(x2_all[1:i], y2_all[1:i], color=:green, lw=1,
                  xlim=(-(l1+l2), l1+l2), ylim=(-(l1+l2), 1),
                  legend=false, aspect_ratio=1, title="Graph")
        scatter!(p2, [x2], [y2], color=:red, markersize=4)

        plot(p1, p2, layout=(1, 2), size=(900, 420))
    end

    gif(anim, save_as, fps=fps)
end

# Params
g = 9.81
l1 = 1.0
l2 = 1.0
m1 = 1.0
m2 = 1.0
tspan = (0.0, 20.0)
dt = 0.01

theta1_0 = pi / 2
omega1_0 = 0.0
theta2_0 = pi / 2
omega2_0 = 0.0

t, theta1, theta2 = rk4_double_pendulum(tspan, dt, theta1_0, omega1_0, theta2_0, omega2_0, g, l1, l2, m1, m2)
animate_double_pendulum(theta1, theta2, l1, l2; save_as="double_pendulum.gif")
```
